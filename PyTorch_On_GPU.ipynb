```python
import time

size=1000
matrix_cup1=torch.randn(size,size)
matrix_cup2=torch.randn(size,size)

start_time=time.time()
result_cpu=torch.matmul(matrix_cup1,matrix_cup2)
cpu_time=time.time()-start_time
print(cpu_time)

matrix_gup1=matrix_cup1.to('cuda')
matrix_gup2=matrix_cup2.to('cuda')
start_time=time.time()
result_gpu=torch.matmul(matrix_gup1,matrix_gup2)
torch.cuda.synchronize()
gpu_time=time.time()-start_time
print("GPU: ",gpu_time)

```
